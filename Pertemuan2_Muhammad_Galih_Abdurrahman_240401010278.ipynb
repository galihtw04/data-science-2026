```python
# Muat library
import pandas as pd
import numpy as np

# Muat Dataset Titanic (menggunakan Link berbeda karena tidak bisa diakses)
URL = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(URL)
print("Dataset berhasil dimuat!")
```

```python
print(df.shape, df.columns.tolist())
print(df.isnull().sum())
```

```python
baris, kolom = df.shape
print(f"Jumlah baris dan kolom dataset Titanic: {baris} baris dan {kolom} kolom")
```

```python
missing = df.isnull().sum()
kolom_missing = missing[missing > 0]
print("Kolom yang memiliki nilai kosong (missing):")
for col, jumlah in kolom_missing.items():
    print(f"  - {col}: {jumlah} nilai kosong")
```

```python
persen_selamat = df['Survived'].mean() * 100
print(f"Persen penumpang yang selamat (Survived=1): {persen_selamat:.2f}%")
```

```python
wk1 = df[(df['Sex']=='female') & (df['Pclass']==1)]
print(wk1['Survived'].mean(), wk1['Age'].mean())

rata_usia = wk1['Age'].mean()
print(f"Rata-rata usia penumpang wanita kelas 1 yang selamat: {rata_usia:.2f} tahun")
```

```python
hasil = df.groupby('Pclass').agg(
    total=('PassengerId','count'),
    selamat=('Survived','mean'),
).round(3)
print(hasil)

kelas_tertinggi = hasil['selamat'].idxmax()
survival_tertinggi = hasil['selamat'].max() * 100
print(f"\nKelas penumpang dengan tingkat keselamatan tertinggi: Kelas {kelas_tertinggi} ({survival_tertinggi:.1f}%)")
print(f"\nHasil groupby('Pclass')['Survived'].mean():")
print(df.groupby('Pclass')['Survived'].mean().round(3))
```
